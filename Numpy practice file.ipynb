```python
##6 stacking and splitting 
```

```python
a=np.array([1,2,3])
b=np.array([4,5,6])

print(np.vstack((a,b))) #vertical  stacking -row wise
print(np.hstack((a,b)))#horizontal stacking -column wise
```

```python
c=np.array([[1,2,3],[4,5,6]])
print(c)
```

```python
split=np.hsplit(c,3) #horizontal split 
for s in split:
    print(s)#using loop 
```

```python
vsplit=np.vsplit(c,2) #vertical split 
for s in vsplit:
    print(s)#using loop 
```
