```python
import matplotlib.pyplot as plt 
#pyplot is collecion of function that make matplotlib work as MATLAB
```

```python
##1 create first chart 
```

```python
# dummy data
x=[1,2,3,4,5]
y=[10,20,25,35,45]
plt.plot(x,y) #line chart 
plt.show() #display chart in jupyter notebook 
```

```python
## 2 customize chart 
```

```python
plt.plot(figuresize=(6,4)) #adjusting for  figure  size 
plt.plot(x,y,color='green',marker='o',linestyle='dashed',linewidth=2,markersize=12)
plt.title('line chart')
plt.xlabel('x axis')
plt.ylabel('y axis')
plt.show() #we can use Matplotlib website for precise chart knowledge 
```

```python
### 3 advance line chart /multiple line & legends 
```

```python
x=[1,2,3,4,5]
y1=[10,20,25,35,45]
y2=[20,35,45,50,55]
plt.plot(x,y1,color='g',marker='o',linestyle='--',linewidth=2,markersize=12,label='sales 2024')
plt.plot(x,y2,color='b',marker='o',linestyle='--',linewidth=2,markersize=12,label='sales 2025')
plt.title('advance line chart')
plt.xlabel('x axis')
plt.ylabel('y axis')
plt.legend() #to show label in legend 
plt.show()
```

```python
### 4  Bar chart 
```

```python
x=['A','B','C','D','E']
y=[10,20,30,40,50]
plt.bar(x,y)
plt.title('bar chart')
plt.xlabel('x axis')
plt.ylabel('y axis')
plt.show()
```

```python
### 5 histogram 
```

```python
#it is used for distribution  analysis 
import random
data=[random.randint(1,10) for _ in range (100)]
plt.hist(data,bins=10)
plt.title('histogram example')
plt.show()
```

```python
###6 pie chart 
```

```python
categories=['A','B','C','D','E']
sales=[10,20,30,40,50]
plt.pie(sales,labels=categories,autopct='%1.1f%%',startangle=90)
plt.title('pie chart')
plt.show()


```

```python
### 7 scatter plot 
#use to find the relation betn variables 
```

```python
#data
y1=[10,20,30,40,50]
y2=[15,25,35,46,55]
plt.scatter(y1,y2)
plt.title('scatter plot')
plt.show()
```

```python
### 8 subplots 
```

```python
#used to show multiple charts in one figure
```

```python
#data1 bar chart
categories=['A','B','C','D','E']
sales=[10,20,30,40,50]
#2 scatter chart
y1=[10,20,30,40,50]
y2=[15,25,35,46,55]
plt.figure(figsize=(6,4))
# first plot bar chart
plt.subplot(1,2,1) #row ,column,position 
plt.bar(categories,sales)
plt.title('daily sales')
plt.xlabel('week days')
plt.ylabel('sales')


plt.subplot(1,2,2) #row ,column,position 
plt.scatter(y1,y2)
plt.title('user example')
plt.xlabel('user 1')
plt.ylabel('user 2')
plt.show

```

```python
### 9 matplotlib with pandas-real data
```

```python
#create dataframe 
import pandas as pd 
#data
data={
    'month':['jan','feb','mar','april','may','jun','july','aug','sep','oct','nov','dec'],
    'sales':[1200,1300,1400,1500,1600,1700,1800,1900,2000,1100,2200,3300]
}
df=pd.DataFrame(data)
df
```

```python
plt.bar(df['month'],df['sales'])
plt.title('matplotlib with pandas')
plt.xlabel('x axis')
plt.ylabel('y axis')
plt.show()
```

```python
### 10 save charts
```

```python
plt.bar(df['month'],df['sales'])
plt.title('matplotlib with pandas')
plt.xlabel('x axis')
plt.ylabel('y axis')
plt.savefig('monthly_sales_from_df.png')
plt.show()
```
