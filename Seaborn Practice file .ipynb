```python
import matplotlib.pyplot as plt 
```

```python
import seaborn as sns 
import pandas as pd 
```

```python
### 1 line plot
var=[1,2,3,4,5,6,7]
var_1=[2,3,4,1,4,7,8]
##seaborn graph 
x_1=pd.DataFrame({'var':var,'var_1':var_1})

sns.lineplot(x='var',y='var_1',data=x_1)
plt.show()
```

```python
#using matplotlib
plt.plot(var,var_1)
plt.show()
```

```python
y_1=sns.load_dataset('penguins') ##dataset from github 
y_1
```

```python
sns.lineplot(x='bill_length_mm',y='flipper_length_mm',data=y_1,hue='sex',size=10,palette='Accent',markers=['o','>'])
plt.show()
```

```python
y_1=sns.load_dataset('penguins').head(20) 
y_1
```

```python
sns.lineplot(x='bill_length_mm',y='bill_depth_mm',data=y_1,hue='sex',
             size=10,palette='Accent',markers=['o','>'])
plt.grid()
plt.title('seaborn line plot ')
plt.show()
```

```python
### 2 Histogram  using  seaborn
```

```python
var=sns.load_dataset('penguins') ##dataset from github 
var
```

```python
sns.displot(var['flipper_length_mm'],bins=[170,180,190,200,210,220,230,240],
           kde=True,rug=True) #bins command use to addjsut tha bins ,#kde is use to show the line on graph 
plt.show()
```

```python
### 3 bar plot in seaborn  
```

```python
var=sns.load_dataset('penguins') ##dataset from github 
var
```

```python
sns.barplot(x=var.island,y=var.bill_length_mm,hue=var.sex)#hue parameter use to separate male and female candidates
plt.show()
```

```python
sns.barplot(x='bill_depth_mm',y='bill_length_mm',hue='sex',data=var,orient="v")
plt.show()
```

```python
sns.barplot(x='bill_depth_mm',y='bill_length_mm',hue='sex',data=var,orient="h")
plt.show()
```

```python
sns.barplot(x='island',y='bill_length_mm',hue='sex',data=var,orient="v",capsize=1,alpha=0.5)
sns.set(style="darkgrid")
plt.show()
```

```python
##count plot in seaborn 
```

```python
var=sns.load_dataset('tips') #dataset from github 
var
```

```python
sns.countplot(x="sex",data=var,hue="smoker",alpha=0.9,saturation=0.1)
plt.show()
```

```python
### violin plot in seaborn 
```

```python
var=sns.load_dataset('tips') #dataset from github 
var
```

```python
sns.violinplot(x='day',y='total_bill',data=var,linewidth=2,saturation=0.1)
plt.show()
```

```python
sns.violinplot(x='day',y='total_bill',data=var,hue='time')
plt.show()
```

```python
sns.violinplot(x='day',y='total_bill',data=var,hue='sex',split=True)
plt.show()
```

```python
sns.violinplot(x='day',y='total_bill',data=var,hue='sex',split=True,inner='quart')
plt.show()
```

```python
sns.violinplot(x=var["total_bill"])
plt.show
```

```python
### pair plot in seabron 
```

```python
var=sns.load_dataset('tips')
var
```

```python
sns.pairplot(var,hue='sex')
plt.show()
```

```python
sns.pairplot(var,vars=['total_bill','tip'],hue='sex',kind='reg',diag_kind='hist')
plt.show()
```

```python
### strip plot in seaborn 
```

```python
var=sns.load_dataset('tips')
var
```

```python
sns.stripplot(x='day',y='total_bill',data=var,hue='sex',linewidth=2)
plt.show()
```
