```python
###10 concatenate and merge DataFrame (JOINS)
df1=pd.DataFrame({'ID':[1,2,3],'name':['A','B','C']})
df1
```

```python
df2=pd.DataFrame({'ID':[1,2,3,4],'scores':[88,96,77,79]})
df2
```

```python
pd.concat([df1,df2],axis=0) #vertical joinn /row level/top on top 
```

```python
pd.concat([df1,df2],axis=1) #horizontal/column level 
```

```python
#merge =join 

pd.merge(df1,df2,how='inner',on='ID')
```

```python
pd.merge(df1,df2,how='inner',left_on='ID',right_on='ID')
```
